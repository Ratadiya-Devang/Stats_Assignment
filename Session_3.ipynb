```python
#  2.
# Write a Python function 
# called calculate_standard_deviation(scores) that takes a list of exam scores and returns the standard deviation rounded to two decimal places.

school1_marks = [100,49,30]

school2_marks = [20,60,70,80,90,100]

def standard_deviation(marks):
    import numpy as np

    total = 0
    m = np.mean(marks)
   
    for i in marks:
        data = np.square(i-m)
        total = total+data


    variance = total/len(marks)
    std = np.sqrt(variance)
    print("variance is :== ",variance)
    print("standard deviation is :== ",std)
    print("\n")
          
       
    

standard_deviation(school1_marks)
standard_deviation(school2_marks)
```
