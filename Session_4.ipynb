```python
import numpy as np 
import matplotlib.pyplot as plt 
import pandas as pd
import scipy.stats as st        
```

```python
# 1.Given the daily order counts for a Swiggy delivery partner over 10 days: [12, 15, 17, 14, 13, 16, 200, 18, 14, 15],
#  plot a simple graph (hand-drawn or using any tool) and describe whether the data is left skewed, right skewed, or symmetrical.
#  Explain your reasoning based on the shape.

order = [12, 15, 17, 14, 13, 16, 200, 18, 14, 15]

days = range(1,11)

plt.plot(days,order)
plt.title("Daily Swigy Orders")
plt.xlabel("Days")
plt.ylabel("Orders")

'''The data is right skewed (positively skewed) because most daily order counts are between 12 and 18, while one unusually high value (200) 
creates a long tail on the right side. This outlier pulls the distribution toward the right.'''
```

```python
# 2.Take the following list of Instagram influencer follower counts (in thousands): [5, 7, 8, 8, 9, 10, 12, 15, 95]. 
# Calculate the mean and median, then state which is higher and what that tells you about the skewness of the data.<br><br><em><strong>Hint:</strong> 
# If the mean is pulled away from the median, the data is skewed in that direction.</em>

follower = [5, 7, 8, 8, 9, 10, 12, 15, 95]
influencer = range(1,10)

mean = np.mean(follower)
median = np.median(follower)

print(f"mean is = {mean}")
print(f"median is = {median}")


plt.plot(influencer,follower,marker='X')
plt.title("Follower Analysis")
plt.xlabel("Influencer")
plt.ylabel("Follower")
plt.grid()

'''that defination according to the data median right side mean was lie so it is the right skewed'''
```
