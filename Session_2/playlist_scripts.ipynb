```python


# Given three lists representing YouTube video views for three different channels, compare the mean, median, and mode for
#  each channel, and decide which channel's data is most affected by outliers.
# <br><br><em><strong>Hint:</strong> Try using one list with a very high value (e.g., [100, 120, 110, 105, 5000])
#  to see the effect on mean vs median.</em>


abc = [100, 120, 110, 105, 5000]




def check_outlier(data):
        
    q1 = np.percentile(data,25)
    q2 = np.percentile(data,50)
    q3 = np.percentile(data,75)
    iqr = q3 -q1 
    upperfence = q1 + (1.5*iqr)
    lowerfence = q3 - (1.5*iqr)




    print(f"q1 is := {q1}")
    print(f"q2 is := {q2}")
    print(f"q3 is := {q3}")
    print(f"iqr is = {iqr}")
    print(f"upper fense is = {upperfence}")
    print(f"lowerfence is = {lowerfence}")


    

    for i in data:
        if i < lowerfence or i > upperfence :
            print(f"outlier is : == {i}" )
        
    plt.boxplot(data)

check_outlier(abc)
        



```
