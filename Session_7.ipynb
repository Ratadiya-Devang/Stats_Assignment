```python

''' 1.Create two Python lists: one representing the number of hours you spend on Instagram each day for a week, and another for the 
 number of posts you make each day. Plot these as a scatter plot using matplotlib and visually describe whether the relationship 
 looks like a positive, negative, or zero correlation.'''


import matplotlib.pyplot as plt 



hour = [2,3,4,5,1]
post = [3,5,7,8,2]

plt.scatter(x=hour,y=post,color='blue',marker='X')

plt.xlabel("hours")
plt.ylabel("post")


print("as per the graph we see when your hour was increase your post was increase")
print("so we will say that the data has positive correlation")
```

```python
''' 2. Given the following data for 6 friends: their monthly Zomato order count and their monthly Swiggy order count, calculate the correlation 
 coefficient between the two lists using numpy's corrcoef(). Print the result and explain if the correlation is positive, negative, or zero.'''

import numpy as np 

zomato_orders = [8, 10, 12, 15, 18, 20]
swiggy_orders = [7, 9, 11, 14, 17, 19]


corelation_matrix = np.corrcoef(zomato_orders,swiggy_orders)
corelation = correlation_matrix[0,1]
 
print(corelation)
if 0 < corelation :
    print("positive corelation")

else :
    print("negative corelation")
```

```python
# 3. Think of two variables from your daily life (for example, number of YouTube videos watched per day and daily study hours).
#  Write a short paragraph describing whether you expect a positive, negative, or zero correlation between them, and why.


youtube_videos = [1, 2, 3, 4, 5, 6, 7]
study_hours = [7, 6, 5, 4, 3, 2, 1]


correlation_matrix=np.corrcoef(youtube_videos,study_hours)

correlation = correlation_matrix[0,1]

if correlation == 0 :
    print("zero correlation")
elif correlation > 0 :
    print("positive correlation")
else :
    print("negative correlation")

print("\nExplanation:")
print("As the number of YouTube videos increases,")
print("the daily study hours decrease.")
print("Therefore, the relationship is Negative Correlation.")
```
