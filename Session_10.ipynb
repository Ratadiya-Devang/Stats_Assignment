```python
# 1. Write a Python function bayes_posterior(prior, likelihood, evidence) that calculates and returns the posterior probability using Bayes theorem 
# formula for given values.

def bayes_posterior(prior, likelihood, evidence):
    
    posterior = (likelihood * prior) / evidence
    return posterior


# Example
prior = 0.30
likelihood = 0.80
evidence = 0.50

result = bayes_posterior(prior, likelihood, evidence)

print("Posterior Probability:", result)
```

```python

# 2.Given the following scenario: On Swiggy, 10% of restaurants are 'premium', and 80% of premium restaurants offer a discount, while only 20% of 
# regular restaurants do. If you see a discount, use Bayes theorem in code to find the probability that the restaurant is premium.<br><br><em><strong>
# Hint:</strong> Define prior, likelihood, and evidence in your code with comments.</em>



prior = 0.10


likelihood = 0.80
regular_discount = 0.20

regular = 0.90


evidence = (likelihood * prior) + (regular_discount * regular)

posterior = (likelihood * prior) / evidence

print("Prior Probability (Premium):", prior)
print("Likelihood (Discount | Premium):", likelihood)
print("Evidence (Discount):", evidence)
print("Posterior Probability (Premium | Discount):", posterior)
```

```python
# 3.Simulate a basic spam detection filter: Given a list of 20 email subjects (some spam, some not), and the probability that 'free' appears in spam 
# vs non-spam emails, write code to calculate the probability an email is spam if it contains 'free', using Bayes theorem.


emails = [
    "Win a free iPhone",
    "Meeting at 3 PM",
    "Free gift for you",
    "Project submission reminder",
    "Claim your free reward",
    "Team meeting tomorrow",
    "Limited time free offer",
    "Invoice attached",
    "Free vacation package",
    "Happy Birthday",
    "Free cashback today",
    "Lunch Invitation",
    "Your bank statement",
    "Free coupons inside",
    "Update your profile",
    "Congratulations! You won",
    "Free movie tickets",
    "Weekly Report",
    "Exclusive free bonus",
    "Doctor Appointment"
]

prior_spam = 10 / 20

likelihood = 0.80


non_spam_likelihood = 0.20


prior_non_spam = 1 - prior_spam


evidence = (likelihood * prior_spam) + (non_spam_likelihood * prior_non_spam)


posterior = (likelihood * prior_spam) / evidence

print("Total Emails:", len(emails))
print("Prior Probability (Spam):", prior_spam)
print("Likelihood P(Free|Spam):", likelihood)
print("Evidence P(Free):", evidence)
print("Posterior Probability P(Spam|Free):", posterior)
```
