```python
# 1. Generate 100 random numbers in Python that follow a normal distribution with mean 50 and standard deviation 10 
# using numpy, then plot the bell curve using matplotlib.


data = np.random.normal(loc=50,scale = 10 , size = 100)
plt.figure(figsize=(8,5))
plt.hist(data,edgecolor="black",color='orange')
plt.show()
```

```python
# 2. Given an array of 20 daily step counts from your fitness tracker app, calculate the mean, median, and mode, and 
# check if they are approximately equal as expected in a normal distribution.<br><br><em><strong>Hint:</strong> Use
# numpy.mean(), numpy.median(), and scipy.stats.mode().</em>
from scipy import stats
steps = [8200, 8500, 7900, 8100, 8400,
         8300, 8600, 8000, 8200, 8450,
         8150, 8350, 8500, 8250, 8100,
         8400, 8300, 8200, 8500, 8150]


mean = np.mean(steps)
median = np.median(steps)
mode_value = stats.mode(steps)
mode = mode_value[0]

print("mean is = ",mean)
print("median is = ",median)
print("mode is = ",mode)

if (mean-mode) < 200 and (mean-median) < 200:
    print("normal Distribution")
else:
    print("not normal Distribution")


```

```python
# 3.Simulate the scores of 200 players in an online game (e.g., PUBG, Free Fire) using a normal distribution with mean
#  70 and standard deviation 15, then count how many players scored within one standard deviation (between 55 and 85) 
# of the mean.<br><br><em><strong>Hint:</strong> Use the 68-95-99 rule to check your answer.</em>

data = np.random.normal(loc=70,scale=15,size=200)
count = np.sum((data>55)&(data<85))

print("length scores",len(data))
print("player scoring between 55 and 85=",count)
print("Percentage :", (count / len(data)) * 100, "%")

```
